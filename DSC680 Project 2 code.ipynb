## Data cleaning

```python
#initialize 

FIRST_AIR_CUTOFF = "2022-12-31"  # shows must have premiered by this date so a renewal decision has had time to happen
MIN_VOTES = 10                   # inclusion filter only
KEEP_TYPES = ["Scripted"]        # focus on scripted shows
```

```python
#data cleaning

shows = movie_db.copy()
shows["first_air_date"] = pd.to_datetime(shows["first_air_date"], errors = "coerce")
shows["last_air_date"] = pd.to_datetime(shows["last_air_date"], errors = "coerce")
 
print("Rows:", len(shows))
print("\nMissing first_air_date:", shows["first_air_date"].isna().sum())
print("\nstatus:\n", shows["status"].value_counts(dropna=False))
print("\ntype:\n", shows["type"].value_counts(dropna=False).head(10))
print("\nnumber_of_seasons:\n", shows["number_of_seasons"].value_counts().sort_index().head(10))
```

```python
#label: "received a second season"

shows = shows[shows["number_of_seasons"] >= 1].copy()  # drop 0-season placeholders
shows["got_second_season"] = (shows["number_of_seasons"] >= 2).astype(int)
```

```python
#eligibility filter 

steps = [("start (>=1 season)", len(shows))]
 
shows = shows[shows["first_air_date"].notna()]
steps.append(("has first_air_date", len(shows)))
 
shows = shows[shows["first_air_date"] <= FIRST_AIR_CUTOFF]
steps.append((f"premiered by {FIRST_AIR_CUTOFF}", len(shows)))
 
shows = shows[shows["type"].isin(KEEP_TYPES)]
steps.append((f"type in {KEEP_TYPES}", len(shows)))
 
shows = shows[shows["vote_count"] >= MIN_VOTES]
steps.append((f"vote_count >= {MIN_VOTES}", len(shows)))
 
#one-season shows still marked as active are undecided, not "not renewed"
undecided = (shows["number_of_seasons"] == 1) & (shows["status"].isin(["Returning Series", "In Production", "Planned", "Pilot"])
                                                 | (shows["in_production"] == True))
print(f"Dropping {undecided.sum():,} one-season shows still active (decision unknown)")

shows = shows[~undecided]
steps.append(("drop undecided one-season shows", len(shows)))

print("\nFilter funnel:")
for name, n in steps:
    print(f"  {name:<40} {n:>9,}")
print(f"\nFinal modeling universe: {len(shows):,} shows")
print(shows["got_second_season"].value_counts())
print(f"\nRenewal rate: {shows['got_second_season'].mean():.1%}")
```

```python
#build_universe() and sensitivity table

active = ["Returning Series", "In Production", "Planned", "Pilot"]
active_tvs = set(tv_series.loc[tv_series["status"].isin(active), "id"])

def build_universe(cutoff, min_votes):
    s = movie_db.copy()
    s["first_air_date"] = pd.to_datetime(s["first_air_date"], errors = "coerce")
    s = s[(s["number_of_seasons"] >= 1) & s["first_air_date"].notna()]
    s = s[(s["first_air_date"] <= cutoff) & (s["type"] == "Scripted") & (s["vote_count"] >= min_votes)]
    undecided = (s["number_of_seasons"] == 1) & (s["status"].isin(active) | (s["in_production"] == True) | s["id"].isin(active_tvs))
    s = s[~undecided].copy()
    s["got_second_season"] = (s["number_of_seasons"] >= 2).astype(int)
    s["primary_country"] = s["origin_country"].fillna("").str.split(",").str[0].str.strip()
    return s

rows = []
for cutoff in ["2018-12-31", "2020-12-31", "2022-12-31"]:
    for mv in [10, 25, 50]:
        s = build_universe(cutoff, mv)
        west = s[s["primary_country"].isin(["US", "GB", "CA", "AU"])]
        rows.append({"cutoff": cutoff, "min_votes": mv, "n": len(s),
                     "renewal_rate": round(s["got_second_season"].mean(), 3),
                     "n_US_GB_CA_AU": len(west),
                     "rate_US_GB_CA_AU": round(west["got_second_season"].mean(), 3)})
print(pd.DataFrame(rows))
```

```python
#one-season status check with 2022 cutoff

one = build_universe("2022-12-31", 10).query("number_of_seasons == 1")
print(one["status"].value_counts())
```

```python
#final modeling universe

FIRST_AIR_CUTOFF = "2020-12-31"
MIN_VOTES = 10

shows = build_universe(FIRST_AIR_CUTOFF, MIN_VOTES)
print(f"{len(shows):,} shows | renewal rate {shows['got_second_season'].mean():.1%}")
```

```python
#label sanity check, one-season shows only

one = shows[shows["number_of_seasons"] == 1]

print(one["status"].value_counts())
print(f"Canceled share of one-season shows: {(one['status'] == 'Canceled').mean():.1%}")
```

## Join

```python
#join the ratings/streaming tables

import re
 
def norm_title(s):
    return re.sub(r"[^a-z0-9]", "", str(s).lower())
 
ts = tv_shows.drop(columns = [c for c in tv_shows.columns if c.startswith("Unnamed")]).copy()
ts["imdb_score"] = pd.to_numeric(ts["IMDb"].astype(str).str.split("/").str[0], errors = "coerce")
ts["rt_score"] = pd.to_numeric(ts["Rotten Tomatoes"].astype(str).str.split("/").str[0], errors = "coerce")
ts["title_key"] = ts["Title"].apply(norm_title)
 
shows["title_key"] = shows["name"].apply(norm_title)
shows["year_key"] = shows["premiere_year"]
 
merged = shows.merge(
    ts[["title_key", "Year", "imdb_score", "rt_score", "Netflix", "Hulu", "Prime Video"]],
    left_on = ["title_key", "year_key"], 
    right_on = ["title_key", "Year"], 
    how = "left")
match_rate = merged["imdb_score"].notna().mean() 

print(f"TMDb shows matched to ratings table: {match_rate:.1%} "
      f"\n({ts.shape[0]:,} ratings rows available)")
```

## Feature engineering

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (ConfusionMatrixDisplay, RocCurveDisplay, accuracy_score, f1_score, roc_auc_score)
from sklearn.preprocessing import StandardScaler
```

```python
TARGET = "got_second_season"
TRAIN_END = 2017  #train on premieres <= 2017
TEST_START = 2018  #test on premieres 2018-2020  
WEST = ["US", "GB", "CA", "AU"]
```

```python
#feature encoding 
 
def collapse(series, keep):
    #keep listed categories, label blanks 'Unknown' and everything else 'Other'
    s = series.fillna("").astype(str).str.strip().replace("", "Unknown")
    return s.where(s.isin(keep) | (s == "Unknown"), "Other")
 
def make_features(df, train_mask, top_networks = 25, top_countries = 12, top_langs = 8, min_genre = 50):
    X = pd.DataFrame(index = df.index)
 
    #numeric launch-time features
    X["premiere_year"] = df["premiere_year"]
    X["premiere_month"] = df["premiere_month"]
    X["n_genres"] = df["n_genres"]
    X["n_networks"] = df["n_networks"]
    X["n_creators"] = df["n_creators"]
    rt = df["episode_run_time"].where(df["episode_run_time"].between(1, 180))   
    X["runtime_missing"] = rt.isna().astype(int)
    X["episode_run_time"] = rt.fillna(rt[train_mask].median())
 
    #genres to multi-hot 
    glist = df["genres"].fillna("").apply(lambda s: [g.strip() for g in s.split(",") if g.strip()])
    gcounts = glist[train_mask].explode().value_counts()
    for g in gcounts[gcounts >= min_genre].index:
        X[f"genre_{g}"] = glist.apply(lambda l: int(g in l))
 
    #high-cardinality categoricals to top-N from TRAIN, rest "Other"
    for col, n, prefix in [("primary_network", top_networks, "net"),
                           ("primary_country", top_countries, "country"),
                           ("original_language", top_langs, "lang")]:
        keep = df.loc[train_mask, col].fillna("").replace("", np.nan).value_counts().head(n).index
        dummies = pd.get_dummies(collapse(df[col], keep), prefix = prefix).astype(int)
        X = pd.concat([X, dummies], axis=1)
    return X
```

```python
#time-based split

train_mask = (shows["premiere_year"] <= TRAIN_END).to_numpy()
test_mask = (shows["premiere_year"] >= TEST_START).to_numpy()
 
X_all = make_features(shows, train_mask)
y_all = shows[TARGET]
 
X_train, y_train = X_all[train_mask], y_all[train_mask]
X_test, y_test = X_all[test_mask], y_all[test_mask]
test_df = shows[test_mask]
 
print(f"Features: {X_all.shape[1]}")
print(f"Train: {len(X_train):,} shows (premiered <= {TRAIN_END}), renewal rate {y_train.mean():.1%}")
print(f"Test:  {len(X_test):,} shows (premiered >= {TEST_START}), renewal rate {y_test.mean():.1%}")
```

```python
#fit and evaluate helpers
 
def fit_models(Xtr, ytr):
    scaler = StandardScaler().fit(Xtr)
    logreg = LogisticRegression(max_iter = 2000, random_state = 42).fit(scaler.transform(Xtr), ytr)
    rf = RandomForestClassifier(n_estimators = 400, max_depth = 10, min_samples_leaf = 10, random_state = 42, n_jobs = -1).fit(Xtr, ytr)
    return scaler, logreg, rf
 
def score(name, proba, y_true, thresh = 0.5):
    pred = (proba >= thresh).astype(int)
    print(f"{name:<22} AUC {roc_auc_score(y_true, proba):.3f} | "
          f"acc {accuracy_score(y_true, pred):.3f} | F1 {f1_score(y_true, pred):.3f}")
```

```python
#main results vs baselines

scaler, logreg, rf = fit_models(X_train, y_train)
lr_proba = logreg.predict_proba(scaler.transform(X_test))[:, 1]
rf_proba = rf.predict_proba(X_test)[:, 1]
 
#baseline 1: always predict the training majority class
maj = int(y_train.mean() >= 0.5)
print(f"Majority-class baseline accuracy: {accuracy_score(y_test, np.full(len(y_test), maj)):.3f}")
 
#baseline 2: country-only renewal rate 
country_rate = shows[train_mask].groupby("primary_country")[TARGET].mean()
base_proba = test_df["primary_country"].map(country_rate).fillna(y_train.mean()).to_numpy()
 
print("\nTEST RESULTS (time-based split):")
score("Country-only baseline", base_proba, y_test)
score("Logistic regression", lr_proba, y_test)
score("Random forest", rf_proba, y_test)
```

```python
#evaluation visuals

fig, ax = plt.subplots(figsize = (10, 10))
RocCurveDisplay.from_predictions(y_test, base_proba, name = "Country-only baseline", ax = ax)
RocCurveDisplay.from_predictions(y_test, lr_proba, name = "Logistic regression", ax = ax)
RocCurveDisplay.from_predictions(y_test, rf_proba, name = "Random forest", ax = ax)

ax.plot([0, 1], [0, 1], "--", color = "gray", label = "Chance")
ax.set_title("ROC Curve Comparison (test: premiered 2018-2020)")
ax.legend()
#plt.savefig("roc_curve.png")
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize = (10, 5))

for ax, name, proba in [(axes[0], "Logistic Regression", lr_proba), (axes[1], "Random Forest", rf_proba)]:
    ConfusionMatrixDisplay.from_predictions(y_test, (proba >= 0.5).astype(int), ax = ax, cmap = "Blues", 
                                            display_labels = ["One season", "2+ seasons"])
    ax.set_title(name)
#plt.savefig("correlation_charts.png")
plt.show()
```

```python
#fairness/subgroup check: does performance differ by region? 

print("Test performance by region (random forest):")
region = np.where(test_df["primary_country"].isin(WEST), "US/GB/CA/AU", "All other")
for g in ["US/GB/CA/AU", "All other"]:
    m = region == g
    yt = y_test.to_numpy()[m]
    if len(np.unique(yt)) > 1:
        print(f"  {g:<12} n = {m.sum():>5} | renewal rate {yt.mean():.1%} | "
              f"AUC {roc_auc_score(yt, rf_proba[m]):.3f} | "
              f"mean predicted {rf_proba[m].mean():.1%}")
```

```python
#robustness test 

print("ROBUSTNESS CHECKS (random forest AUC on test):")
 
#drop premiere_year: does the model rely on the downward time trend?
no_year = [c for c in X_all.columns if c != "premiere_year"]
rf_ny = RandomForestClassifier(n_estimators = 400, max_depth = 10, min_samples_leaf = 10, random_state = 42, 
                               n_jobs = -1).fit(X_train[no_year], y_train)

print(f"Without premiere_year: {roc_auc_score(y_test, rf_ny.predict_proba(X_test[no_year])[:, 1]):.3f}")
```

```python
#canceled-only negatives: negatives = one-season shows TMDb marks 'Canceled'

keep = (shows[TARGET] == 1) | (shows["status"] == "Canceled")
Xc, yc = X_all[keep.to_numpy()], y_all[keep]
trc, tec = train_mask[keep.to_numpy()], test_mask[keep.to_numpy()]
rf_c = RandomForestClassifier(n_estimators = 400, max_depth = 10, min_samples_leaf = 10, class_weight = "balanced", 
                              random_state = 42, n_jobs = -1).fit(Xc[trc], yc[trc])

print(f"Canceled-only negatives: {roc_auc_score(yc[tec], rf_c.predict_proba(Xc[tec])[:, 1]):.3f} "
      f"\n(n_test={tec.sum()}, positive share {yc[tec].mean():.1%})")
```

```python
#western shows only

w = shows["primary_country"].isin(WEST).to_numpy()
rf_w = RandomForestClassifier(n_estimators = 400, max_depth = 10, min_samples_leaf = 10, random_state = 42, 
                              n_jobs = -1).fit(X_all[train_mask & w], y_all[train_mask & w])

print(f"US/GB/CA/AU shows only: {roc_auc_score(y_all[test_mask & w], rf_w.predict_proba(X_all[test_mask & w])[:, 1]):.3f}")
```

```python
#bootstrap 95% CIs for auc and auc diff

rng = np.random.default_rng(42)
yt = y_test.to_numpy()
preds = {"country": base_proba, "logreg": lr_proba, "rf": rf_proba}
n = len(yt)
boot = {k: [] for k in preds}

for _ in range(1000):
    idx = rng.integers(0, n, n)
    if len(np.unique(yt[idx])) < 2:
        continue
    for k, v in preds.items():
        boot[k].append(roc_auc_score(yt[idx], v[idx]))

boot = {k: np.array(v) for k, v in boot.items()}

for k, v in boot.items():
    print(f"{k:<8} AUC 95% CI: {np.percentile(v, 2.5):.3f} - {np.percentile(v, 97.5):.3f}")

for a, b in [("rf", "country"), ("logreg", "country"), ("rf", "logreg")]:
    d = boot[a] - boot[b]
    print(f"{a} - {b}: median {np.median(d):+.3f}, 95% CI {np.percentile(d, 2.5):+.3f} to {np.percentile(d, 97.5):+.3f}")
```

```python
#grouped permutation importance

cols = list(X_all.columns)
groups = {
    "Region (country + language)": [c for c in cols if c.startswith(("country_", "lang_"))],
    "Network": [c for c in cols if c.startswith("net_")],
    "Genre": [c for c in cols if c.startswith("genre_")] + ["n_genres"],
    "Timing (year, month)": ["premiere_year", "premiere_month"],
    "Format (runtime, creators, network count)": ["episode_run_time", "runtime_missing", "n_creators", "n_networks"]}

base_auc = roc_auc_score(y_test, rf_proba)
prng = np.random.default_rng(0)
rows = []

for name, gcols in groups.items():
    drops = []
    for _ in range(10):
        Xp = X_test.copy()
        Xp[gcols] = X_test[gcols].to_numpy()[prng.permutation(len(Xp))]
        drops.append(base_auc - roc_auc_score(y_test, rf.predict_proba(Xp)[:, 1]))
    rows.append({"feature group": name, "n_cols": len(gcols), "AUC drop": round(np.mean(drops), 3)})

print(pd.DataFrame(rows).sort_values("AUC drop", ascending = False).to_string(index = False))
```
